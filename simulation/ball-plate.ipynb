```python
import mujoco
import mediapy as media
import numpy as np

model = mujoco.MjModel.from_xml_path("generated/ball_plate.xml")
data = mujoco.MjData(model)
mujoco.mj_resetData(model, data)
x_cmd = np.deg2rad(0)
y_cmd = np.deg2rad(0)

data.joint("servo_x_joint").qpos[0] = x_cmd
data.joint("servo_y_joint").qpos[0] = y_cmd

data.actuator("servo_x").ctrl[0] = x_cmd
data.actuator("servo_y").ctrl[0] = y_cmd

mujoco.mj_forward(model, data)

n_frames = 120
height = 300
width = 300
frames = []
# Simulate and display video.
cam = mujoco.MjvCamera()

cam.lookat[:] = [0, 0, 0.04]
cam.distance = 0.35
cam.azimuth = 135
cam.elevation = -25
with mujoco.Renderer(model, height, width) as renderer:
# random initial rotational velocity:
  
#   data.qvel[2:4] = .1 * (np.random.rand(2)-0.5)   
  for i in range(n_frames):
    while data.time < i/30.0:
      mujoco.mj_step(model, data)
    renderer.update_scene(data, cam)
    frame = renderer.render()
    frames.append(frame)
    # if i % 4 == 0:
    #     print(
    #         "plate_x:", np.rad2deg(data.joint("plate_x").qpos[0]),
    #         "plate_y:", np.rad2deg(data.joint("plate_y").qpos[0]),
    #         "servo_x:", np.rad2deg(data.joint("servo_x_joint").qpos[0]),
    #         "servo_y:", np.rad2deg(data.joint("servo_y_joint").qpos[0]),
    #     )

media.show_video(frames, fps=30)
```
